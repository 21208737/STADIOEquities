# Feature Engineering

```python
# AgeGroup
def create_age_group(age):
    if age < 30:
        return "Young"
    elif age < 45:
        return "Adult"
    elif age < 60:
        return "Middle-aged"
    else:
        return "Senior"

df["AgeGroup"] = df["Age"].apply(create_age_group)

df[["Age", "AgeGroup"]].head(10)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="AgeGroup", data=df)

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.show()
```

```python
# Engaged
df["EngagementRisk"] = (
    (df["IsActiveMember"] == 0).astype(int)
    + (df["NumOfProducts"] <= 1).astype(int)
    + (df["Complain"] == 1).astype(int)
)

df[["IsActiveMember", "EngagementRisk"]].head(10)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="EngagementRisk", data=df)

plt.title("Customer Engagement Risk Distribution")
plt.xlabel("Engagement Risk Score")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.show()
```

```python
# Tenure_Age_Ratio
df["Tenure_Age_Ratio"] = (
    df["Tenure"] / df["Age"].replace(0, np.nan)
)

df[["Tenure", "Tenure_Age_Ratio"]].head(10)
```

```python
plt.figure(figsize=(7, 5))

sns.histplot(df["Tenure_Age_Ratio"].dropna(), bins=30, kde=True)

plt.title("Distribution of Tenure-to-Age Ratio")
plt.xlabel("Tenure-to-Age Ratio")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.show()
```

```python
# Balance_Salary_Ratio
df["Balance_Salary_Ratio"] = (
    df["Balance"] /
    df["EstimatedSalary"].replace(0, np.nan)
)

df[["Balance", "Balance_Salary_Ratio"]].head(10)
```

```python
plt.figure(figsize=(7, 5))

sns.histplot(df["Balance_Salary_Ratio"].dropna(), bins=30, kde=True)

plt.title("Distribution of Balance-to-Salary Ratio")
plt.xlabel("Balance-to-Salary Ratio")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.show()
```

```python
# ZeroBalance
df["ZeroBalance"] = (df["Balance"] == 0).astype(int)
df[["Balance", "ZeroBalance"]].head(10)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="ZeroBalance", data=df)

plt.title("Customers with Zero Balance")
plt.xticks([0, 1], ["Has Balance", "Zero Balance"])
plt.xlabel("Balance Status")
plt.ylabel("Number of Customers")
plt.tight_layout()

plt.show()
```
