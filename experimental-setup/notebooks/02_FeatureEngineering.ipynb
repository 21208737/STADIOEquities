```python
# AgeGroup
```

```python
def create_age_group(age):
    if age < 30:
        return "Young"
    elif age < 45:
        return "Adult"
    elif age < 60:
        return "Middle-aged"
    else:
        return "Senior"

df["AgeGroup"] = df["Age"].apply(create_age_group)
```

```python
df[["Age", "AgeGroup"]].head(10)
```
